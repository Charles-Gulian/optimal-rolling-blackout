```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
# From ICE calculator 2.0 study
time_ice = np.array([1/12, 2, 8, 24])
voll_ice = np.array([20.59, 5.03, 3.07, 2.18])
df_voll = pd.Series(index=time_ice, data=voll_ice)
```

```python
# VoLL model to be fit
time = np.arange(25)
def voll_model(time, v, gamma, beta):
    x = (1 - (1 - gamma)**time) / gamma
    voll = v * np.ones_like(time) + beta * x
    return voll
```

```python
def error(v1, v2):
    return np.sum((v1 - v2)**2)
```

```python
# Model parameters
v = 22
gamma = 0.6
beta = -12

print(v + beta / gamma)
assert v + beta / gamma > 0
```

```python
print(error(voll_ice, voll_model(time_ice, v, gamma, beta)))

plt.scatter(time_ice, voll_ice, label="ICE 2.0")
plt.scatter(time_ice, voll_model(time_ice, v, gamma, beta), color="orange", label="Model")
plt.plot(time, voll_model(time, v, gamma, beta), color="orange")
plt.xlabel("H")
plt.ylabel("VoLL")
plt.show()
```
